```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, mean_squared_error,euclidean_distances
```

```python
data = pd.read_csv("tmdb_5000_movies.csv.xls")
df1 = pd.DataFrame(data)
#df1
```

```python
print(df1.head())
print(df1.info())
print(df1.describe())
print(df1.isnull().sum())
```

```python
df = df1.copy()
df["runtime"] = df["runtime"].fillna(df["runtime"].median())
df["runtime"].isnull().sum()
```

```python
df=df.drop(['tagline'],axis=1)
df.isnull().sum()
df.columns
```

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, r2_score
x = df[['popularity','runtime','vote_count']].values
y = df["vote_average"].values
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
model = KNeighborsRegressor(n_neighbors=3)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
r2 = r2_score(y_test, y_pred)
print("The model predictions :", y_pred[0])
print("The model Accuracy Score: ", r2*100)
```

```python
x = df[['popularity','runtime','vote_count']]
scaler = StandardScaler()
scaled = scaler.fit_transform(x)
model = NearestNeighbors(n_neighbors=3, metric="euclidean")
model.fit(scaled)
target_index = 0
distances, indices = model.kneighbors([scaled[target_index]])
print(f"\nRecommendations for '{df.loc[target_index, 'title']}':")
for idx in indices[0][1:]:
    print(f"- {df.loc[idx, "title"]} (Distance: {distances[0][2]:.2f})")
```
